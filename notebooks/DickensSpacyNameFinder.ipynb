```python
# Initial import
from utils import *
# Reload the module to apply changes
import importlib
import utils
importlib.reload(utils)
from utils import *

import spacy
from datetime import datetime
import csv
```

```python
# load and prepare text into paragraph chunks using functions found in utils.py
targetText = "little-dorrit"
text = get_text(f"../data/{targetText}.txt")
paragraphs = prepare_text(text)
```

```python
# load the spaCy english model
nlp = spacy.load('en_core_web_sm')

# extract all PERSON entities from the text
resultNames = []
for para in paragraphs:
  doc = nlp(para)
  for ent in doc.ents:
    if ent.label_ == "PERSON":
      resultNames.append(ent.text)
```

```python
# remove duplicates and sort the list
uniqueResultNames = list(set(resultNames))
uniqueResultNames.sort()
```

```python
# Generate date stamp
date_stamp = datetime.now().strftime("%Y-%m-%d")

# Define output file name
output_file = f"../output/rawNames_{targetText}_{date_stamp}.csv"

# Save resultNames to CSV
with open(output_file, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Name"])
    for name in uniqueResultNames:
        writer.writerow([name])

print(f"Results saved to {output_file}")
```
